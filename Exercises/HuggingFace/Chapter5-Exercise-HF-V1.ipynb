```python
from datasets import load_dataset
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00462/drugsCom_raw.zip"
```

```python
dataset = load_dataset("csv", data_files=url, delimiter="\t")
```

```python
dataset = dataset.filter(lambda x: x["condition"] is not None)
```

```python
def compute_length(example):
  return {"length": len(example["review"].split())}

dataset = dataset.map(compute_length)
```

```python
top_reviews = dataset.filter(lambda x: x["rating"] > 8 and x["length"] > 30)
```

```python
print(f"Number of top reviews: {len(top_reviews['train'])}")
print("Sample review text:")
print(top_reviews["train"][0]["review"])
```

```python
def convert_drugname_to_lowercase(example):
    return {"drugName": example["drugName"].lower()}
```

```python
dataset_lowercase = dataset.map(convert_drugname_to_lowercase)

print("\n💊 Drug names converted to lowercase:")
for i, row in enumerate(dataset_lowercase["train"].select(range(5))):
    print(f"{i+1}. {row['drugName']}")
```
