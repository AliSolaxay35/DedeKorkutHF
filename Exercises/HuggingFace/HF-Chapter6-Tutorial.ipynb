```python
from transformers import AutoTokenizer
from datasets import load_dataset
```

```python
persian_texts = [
    "سلام، من در حال یادگیری هاگینگ فیس هستم.",
    "کتابخانه توکنایزر بسیار سریع و قدرتمند است.",
    "آموزش مدل‌های زبانی بزرگ نیاز به داده‌های تمیز دارد.",
    "هوش مصنوعی آینده تکنولوژی را تغییر خواهد داد."
] * 100
```

```python
print("Loading lightweight tokenizer: DistilGPT-2 (33% smaller than GPT-2)")
old_tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
```

```python
new_tokenizer = old_tokenizer.train_new_from_iterator(persian_texts, 5000)
```

```python
example = "یادگیری ماشین"

print("\n" + "="*60)
print("📊 TOKENIZATION COMPARISON:")
print("="*60)
print(f"Example text: '{example}'\n")
print(f"Old DistilGPT-2 Tokens: {old_tokenizer.tokenize(example)}")
print(f"New Persian Tokens: {new_tokenizer.tokenize(example)}")
```

```python
test_word = ["یادگیری", "هاگینگ", "ماشین", "سلام"]
print("\n" + "="*60)
print("📝 WORD-LEVEL EFFICIENCY COMPARISON:")
print("="*60)

for word in test_word:
  old_tokens = old_tokenizer.tokenize(word)
  new_token = new_tokenizer.tokenize(word)
  print(f"\nWord: '{word}'")
  print(f"  → Old tokenizer: {len(old_tokens)} tokens {old_tokens}")
  print(f"  → New tokenizer: {len(new_token)} tokens {new_token}")
```
