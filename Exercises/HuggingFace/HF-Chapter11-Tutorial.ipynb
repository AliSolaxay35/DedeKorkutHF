```python
from transformers import AutoTokenizer
```

```python
model_id = "HuggingFaceTB/SmolLM2-135M-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_id)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

print(f"✅ Model loaded: {model_id}")
print(f"✅ Tokenizer vocabulary size: {len(tokenizer)}")
```

```python
messages = [
    {"role": "system", "content": "تو یک دستیار هوشمند هستی که به زبان فارسی پاسخ می‌دهی."},
    {"role": "user", "content": "پایتون چیست؟"},
    {"role": "assistant", "content": "پایتون یک زبان برنامه‌نویسی سطح بالا و محبوب است."},
    {"role": "user", "content": "چه مزایایی دارد؟"}
]

print("📝 Original messages:")
for msg in messages:
  print(f"  {msg['role']}: {msg['content']}")
```

```python
formatted_chat = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=False
)

print("\n--- Formatted Chat Template ---")
print(formatted_chat)
print("\n" + "="*60)
```

```python
print("\n📝 EXERCISE FOR YOU:")
print("="*60)
print("Write a function that takes a list of simple dictionaries")
print("(containing 'question' and 'answer') and converts them to")
print("the standard messages format shown above.")
print("\nAssume the system role is always:")
print("\"You are a helpful math tutor\"")
```

```python
def convert_qa_to_messages(qa_list, system_prompt="You are a helpful math tutor"):
  output_messages = []
  output_messages.append({"role": "system", "content": system_prompt})
  for qa in qa_list:
    output_messages.append({"role": "user", "content": qa["question"]})
    output_messages.append({"role": "assistant", "content": qa["answer"]})

  return output_messages
```

```python
qa_data = [
    {"question": "What is 2+2?", "answer": "2+2 equals 4."},
    {"question": "What is the square root of 9?", "answer": "The square root of 9 is 3."},
    {"question": "Solve for x: 2x + 3 = 7", "answer": "x = 2"}
]

print("\n📊 Original Q&A Data:")
for item in qa_data:
  print(f"  Q: {item['question']}")
  print(f"  A: {item['answer']}")
```

```python
converted_messages = convert_qa_to_messages(qa_data)
print("\n✅ Converted to Messages Format:")
for msg in converted_messages:
  print(f"  {msg['role']}: {msg['content']}")
```

```python
formatted_converted = tokenizer.apply_chat_template(
    converted_messages,
    add_generation_prompt=True,
    tokenize=False
)

print("\n--- Formatted Output (with template) ---")
print(formatted_converted)
```
