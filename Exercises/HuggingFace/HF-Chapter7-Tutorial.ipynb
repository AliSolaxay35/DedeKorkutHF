```python
from transformers import pipeline
```

```python
ner_pipeline = pipeline("ner", model="dbmdz/bert-large-cased-finetuned-conll03-english", aggregation_strategy="simple")
```

```python
text = "Hugging Face is a company based in New York City, founded by Clement."
results = ner_pipeline(text)

print("--- Named Entity Recognition Results ---")
for entity in results:
  print(f"Entity: {entity['word']} | Type: {entity['entity_group']} | Score: {entity['score']:.4f}")
```

```python
mask_filter = pipeline("fill-mask", model="distilbert-base-uncased")
mask_text =  "The capital of France is [MASK]."
mask_results = mask_filter(mask_text)

print("\n--- Fill-Mask Results ---")
for res in mask_results:
    print(f"Prediction: {res['sequence']} | Score: {res['score']:.4f}")
```

```python
test_sentence = "Manus is an AI agent developed in San Francisco"
ner_results = ner_pipeline(test_sentence)

print("\n--- NER Results for 'Manus' Test ---")
for entity in ner_results:
  print(f"Entity: {entity['word']} | Type: {entity['entity_group']} | Score: {entity['score']:.4f}")
```

```python
print("\n" + "="*60)
print("🔍 ANALYSIS:")
print("="*60)
```

```python
manus_detected = False
for entity in ner_results:
  if "Manus" in entity['word']:
    manus_detected = True
    print(f"✅ 'Manus' detected as: {entity['entity_group']} (Score: {entity['score']:.4f})")
    break

if not manus_detected:
  print("❌ 'Manus' was NOT detected as an entity by the model.")
  print("   → This is because the model was trained on CoNLL-03 dataset")
  print("   which doesn't contain 'Manus' in its training data.")
```

```python
print("\n📊 Other entities detected:")
for entity in ner_results:
  if "Manus" not in entity['word']:
    print(f"Entity: {entity['word']} | Type: {entity['entity_group']} | Score: {entity['score']:.4f}")
```
