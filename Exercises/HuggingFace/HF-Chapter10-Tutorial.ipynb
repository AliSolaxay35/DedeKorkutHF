```python
from datasets import Dataset
```

```python
data = {
    "text": [
        "عالی بود حتما پیشنهاد میکنم",
        "افتضاح بود وقتتون رو تلف نکنید",
        "بد نبود ولی میتونست بهتر باشه"
    ],
    "label": [0, 0, 1]
}

ds = Dataset.from_dict(data)

print("📊 Original Dataset:")
print(ds)
print(f"Labels: {ds['label']}")
```

```python
def correct_labels(example):
  if "عالی" in example["text"] and example["label"] == 0:
    example["label"] = 1
  return example

ds_cleaned = ds.map(correct_labels)

print("\n✅ Cleaned Dataset (after label correction):")
print(ds_cleaned)
print(f"Labels: {ds_cleaned['label']}")  # Output should be [1, 0, 1]
```

```python
print("\n" + "="*60)
print("📝 EXERCISE FOR YOU:")
print("="*60)
print("Write a function that removes any text with less than 3 words")
print("from the dataset (filter out low-value data).")
print("\n💡 Hint: Use the .filter() method with a lambda function")
print("   or a custom function that checks word count.")
```

```python
def filter_short_texts(example):
  word_count = len (example["text"].split())
  return word_count >= 3
```

```python
ds_filtered = ds_cleaned.filter(filter_short_texts)

print("\n📊 Dataset after filtering short texts:")
print(ds_filtered)
print(f"Labels: {ds_filtered['label']}")
print(f"Texts: {ds_filtered['text']}")

print("\n" + "="*60)
print("🧪 TESTING THE FILTER:")
print("="*60)
```

```python
data_with_short = {
    "text": [
        "عالی بود حتما پیشنهاد میکنم",
        "افتضاح بود وقتتون رو تلف نکنید",
        "بد نبود ولی میتونست بهتر باشه",
        "بد"  # Short text (only 1 word)
    ],
    "label": [1, 0, 1, 0]
}

ds_test = Dataset.from_dict(data_with_short)

print("Before filtering:")
print(f"Texts: {ds_test['text']}")
print(f"Labels: {ds_test['label']}")

ds_test_filtered = ds_test.filter(filter_short_texts)

print("\nAfter filtering (texts with 3+ words):")
print(f"Texts: {ds_test_filtered['text']}")
print(f"Labels: {ds_test_filtered['label']}")
print(f"✅ Removed {len(ds_test) - len(ds_test_filtered)} short text(s)")
```
