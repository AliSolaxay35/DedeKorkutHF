```python
from transformers import AutoTokenizer
```

```python
turkish_texts = [
    "Merhaba, ben Hugging Face öğreniyorum.",
    "Tokenizasyon kütüphanesi çok hızlı ve güçlüdür.",
    "Büyük dil modellerini eğitmek için temiz veri gerekir.",
    "Yapay zeka gelecekte teknolojiyi değiştirecek."
] * 100
```

```python
print("Loading lightweight tokenizer: DistilGPT-2")
old_tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
```

```python
new_tokenizer = old_tokenizer.train_new_from_iterator(turkish_texts, 52000)
```

```python
example = "makine öğrenmesi"
print("\n" + "="*60)
print("📊 TOKENIZATION COMPARISON (TURKISH):")
print("="*60)
print(f"Example text: '{example}'\n")
print(f"Old DistilGPT-2 Tokens: {old_tokenizer.tokenize(example)}")
print(f"New Turkish Tokens: {new_tokenizer.tokenize(example)}")
```

```python
test_word = ["merhaba", "öğrenme", "yapay", "zeka", "HuggingFace"]
print("\n" + "="*60)
print("📝 WORD-LEVEL EFFICIENCY COMPARISON (TURKISH):")
print("="*60)

for word in test_word:
  old_tokens = old_tokenizer.tokenize(word)
  new_tokens = new_tokenizer.tokenize(word)
  print(f"\nWord: '{word}'")
  print(f"  → Old tokenizer: {len(old_tokens)} tokens {old_tokens}")
  print(f"  → New tokenizer: {len(new_tokens)} tokens {new_tokens}")
```
