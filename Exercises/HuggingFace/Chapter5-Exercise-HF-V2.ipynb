```python
import pandas as pd
from datasets import load_dataset
```

```python
data = {
    "patient_id": [101, 102, 103, 104, 105],
    "drugName": ["Aspirin", "Panadol", "Amoxicillin", "Metformin", "Lipitor"],
    "condition": ["Headache", "Fever", "Infection", "Diabetes", "Cholesterol"],
    "review": [
        "It was excellent and worked very fast for my headache.",
        "Not very effective, I still feel a bit feverish after taking it.",
        "The infection went away in three days. Highly recommended for everyone.",
        "Helps manage my blood sugar levels effectively every day.",
        "Good for cholesterol but had some minor side effects like muscle pain."
    ],
    "rating": [9.5, 4.0, 10.0, 9.0, 7.5]
}
```

```python
df = pd.DataFrame(data)
df.to_csv("my_local_data.csv", index=False)
print("✅ 'my_local_data.csv' file created successfully.")
```

```python
local_dataset = load_dataset("csv", data_files="my_local_data.csv")
print(local_dataset)
```

```python
def compute_review_length(example):
    return {"review_length": len(example["review"].split())}

load_dataset = local_dataset.map(compute_review_length)
```

```python
high_rating_dataset = local_dataset.filter(lambda x: x["rating"] > 8)
```

```python
print("\n🏆 High-rated reviews (filtered):")
```

```python
high_rating_dataset_with_length = high_rating_dataset.map(compute_review_length)

for row in high_rating_dataset_with_length["train"]:
    print((f"- Drug {row['drugName']}: Rating {row['rating']} (Word count: {row['review_length']})"))
```
