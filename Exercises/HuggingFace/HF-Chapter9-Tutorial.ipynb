```python
import gradio as gr
from transformers import pipeline
```

```python
print("="*60)
print("🌍 Loading Translation Models...")
print("="*60)

from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

print("✅ Loading English to German model...")
model_de = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-de")
tokenizer_de = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-de")

print("✅ Loading English to French model...")
model_fr = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-fr")
tokenizer_fr = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-fr")

print("✅ Loading English to Spanish model...")
model_es = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-es")
tokenizer_es = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-es")

print("✅ Loading English to Italian model...")
model_it = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-it")
tokenizer_it = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-it")

print("\n✅ All models loaded successfully!")
```

```python
def translate_to_german(text):
  if not text.strip():
    return "Bitte geben Sie einen Text zum Übersetzen ein."
  inputs = tokenizer_de(text, return_tensors="pt", turncation=True, max_length=512)
  outputs = model_de.generate(**inputs)
  return tokenizer_de.decode(outputs[0], skip_special_tokens=True)

def translate_to_french(text):
    if not text.strip():
        return "Veuillez entrer un texte à traduire."
    inputs = tokenizer_fr(text, return_tensors="pt", truncation=True, max_length=512)
    outputs = model_fr.generate(**inputs)
    return tokenizer_fr.decode(outputs[0], skip_special_tokens=True)

def translate_to_spanish(text):
    if not text.strip():
        return "Por favor ingrese un texto para traducir."
    inputs = tokenizer_es(text, return_tensors="pt", truncation=True, max_length=512)
    outputs = model_es.generate(**inputs)
    return tokenizer_es.decode(outputs[0], skip_special_tokens=True)

def translate_to_italian(text):
    if not text.strip():
        return "Per favore inserisci un testo da tradurre."
    inputs = tokenizer_it(text, return_tensors="pt", truncation=True, max_length=512)
    outputs = model_it.generate(**inputs)
    return tokenizer_it.decode(outputs[0], skip_special_tokens=True)

print("✅ Translation functions created successfully!")
```

```python
german_interface = gr.Interface(
    fn=translate_to_german,
    inputs=gr.Textbox(lines=4, label="Enter English text", placeholder="Hello, how are you?"),
    outputs=gr.Textbox(label="German Translation"),
    title="🇩🇪 English → German",
    description="Translate from English to German using Helsinki-NLP/opus-mt-en-de",
    examples=[
        ["Hello, how are you?"],
        ["I love learning new languages."],
        ["The weather is beautiful today."]
    ],
    theme="huggingface"
)

french_interface = gr.Interface(
    fn=translate_to_french,
    inputs=gr.Textbox(lines=4, label="Enter English text", placeholder="Hello, how are you?"),
    outputs=gr.Textbox(label="French Translation"),
    title="🇫🇷 English → French",
    description="Translate from English to French using Helsinki-NLP/opus-mt-en-fr",
    examples=[
        ["Hello, how are you?"],
        ["I love learning new languages."],
        ["The weather is beautiful today."]
    ],
    theme="huggingface"
)

spanish_interface = gr.Interface(
    fn=translate_to_spanish,
    inputs=gr.Textbox(lines=4, label="Enter English text", placeholder="Hello, how are you?"),
    outputs=gr.Textbox(label="Spanish Translation"),
    title="🇪🇸 English → Spanish",
    description="Translate from English to Spanish using Helsinki-NLP/opus-mt-en-es",
    examples=[
        ["Hello, how are you?"],
        ["I love learning new languages."],
        ["The weather is beautiful today."]
    ],
    theme="huggingface"
)

italian_interface = gr.Interface(
    fn=translate_to_italian,
    inputs=gr.Textbox(lines=4, label="Enter English text", placeholder="Hello, how are you?"),
    outputs=gr.Textbox(label="Italian Translation"),
    title="🇮🇹 English → Italian",
    description="Translate from English to Italian using Helsinki-NLP/opus-mt-en-it",
    examples=[
        ["Hello, how are you?"],
        ["I love learning new languages."],
        ["The weather is beautiful today."]
    ],
    theme="huggingface"
)

print("✅ All interfaces created successfully!")
```

```python
multi_demo = gr.TabbedInterface(
    [
        german_interface,
        french_interface,
        spanish_interface,
        italian_interface,
    ],
    tab_names=[
        "🇩🇪 German",
        "🇫🇷 French",
        "🇪🇸 Spanish",
        "🇮🇹 Italian",
    ]
)

print("\n" + "="*60)
print("🚀 Launching Multi-Language Translator...")
print("="*60)
print("✅ Available languages: German, French, Spanish, Italian")
print("💡 Share the public link to use from anywhere!")
print("="*60)

multi_demo.launch(share=True)
```
